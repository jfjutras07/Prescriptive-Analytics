The selected criteria will be weighted using the Analytic Hierarchy Process (AHP) based on stakeholder judgments.

## 3-Analytic Hierarchy Process (AHP) Criteria Weighting

The Analytic Hierarchy Process (AHP) is used to determine the relative importance of the selected decision criteria.

Each participant evaluates criteria through pairwise comparisons using the Saaty fundamental scale. This approach allows stakeholders to express the relative importance of one criterion compared with another.

The comparison scale is defined as follows:

| Value | Interpretation |
|---|---|
| 1 | Equal importance |
| 3 | Moderate importance |
| 5 | Strong importance |
| 7 | Very strong importance |
| 9 | Extreme importance |

Intermediate values (2, 4, 6, and 8) may be used when a judgment falls between two levels.

Individual comparison matrices are collected from each participant. These judgments are then aggregated using the geometric mean to obtain a collective stakeholder comparison matrix.

The aggregated matrix is used to calculate the final criteria weights and evaluate the consistency of stakeholder judgments.



## 4-Individual Pairwise Comparisons

Each stakeholder independently compares the selected decision criteria using the AHP pairwise comparison approach. Individual judgments are expressed according to the Saaty fundamental scale and recorded as reciprocal comparison matrices.

The objective is to capture expert preferences while ensuring that each participant provides an independent assessment before group aggregation.

The individual matrices generated during the workshop will subsequently be combined using the geometric mean to obtain a collective decision matrix.

**Results Presentation**

Pairwise comparisons follow the Saaty fundamental scale. Values greater than 1 indicate that the row criterion is more important than the column criterion, whereas reciprocal values (1/2, 1/3, etc.) indicate that the column criterion is considered more important. 

For example, in Participant 1's matrix, the value of 3 comparing Patient Coverage (row) with Financial Efficiency (column) indicates that the participant judged patient coverage to be moderately more important than financial efficiency. Conversely, the reciprocal value of 1/3 appears in the opposite position, preserving matrix consistency.

## 5-Aggregate Stakeholder Judgments

## 6-Calculate Criteria Weights

The relative importance of each decision criterion is calculated from the aggregated comparison matrix using the normalized priority vector approach. The calculation process follows three main steps:

**Step 1 — Aggregated Comparison Matrix**

Individual stakeholder judgments are first combined into a single pairwise comparison matrix using the geometric mean. For example, the comparison between Financial Efficiency and Patient Coverage combines the six participant judgments:

(1/3 × 1/4 × 1 × 2 × 1/2 × 1/5)^(1/6) = 0.505

This result indicates that, collectively, stakeholders considered Patient Coverage more important than Financial Efficiency. The aggregated matrix represents the collective stakeholder judgment structure used to derive the final criteria weights.

**Step 2 — Priority Vector Calculation**

The aggregated comparison matrix is converted into a priority vector by normalizing the comparison matrix and calculating the average contribution of each criterion.

First, each value in the comparison matrix is normalized by dividing it by the total value of its corresponding column. For example, the Financial Efficiency column of the aggregated matrix is normalized by dividing each comparison value by the column total. The comparison between Patient Coverage and Financial Efficiency is calculated as:

Patient Coverage / Financial Efficiency = 2.04 / 6.43 = 0.32

This process is repeated for all cells of the comparison matrix, producing a normalized comparison matrix. The average value of each row is then calculated. This row average represents the estimated relative importance of each criterion and forms the priority vector. Repeating this calculation for all criteria produces the following priority vector:

| Criterion | Priority Value |
|---|---:|
| Financial Efficiency | 0.16 |
| Patient Coverage | 0.34 |
| Operational Robustness | 0.24 |
| Implementation Feasibility | 0.08 |
| Workforce Sustainability | 0.18 |

The resulting priority vector represents the normalized importance of each criterion. The weights sum to one and can therefore be directly applied in the subsequent MCDA evaluation.

**Step 3 — Weight Interpretation**

The normalized weights represent the contribution of each criterion to the final decision model.

| Criterion | Weight | Interpretation |
|---|---:|---|
| Patient Coverage | 0.34 | Represents 34% of the decision importance |
| Operational Robustness | 0.24 | Represents 24% of the decision importance |
| Workforce Sustainability | 0.18 | Represents 18% of the decision importance |
| Financial Efficiency | 0.16 | Represents 16% of the decision importance |
| Implementation Feasibility | 0.08 | Represents 8% of the decision importance |

These weights represent stakeholder priorities that will be applied in the subsequent TOPSIS evaluation of workforce planning scenarios.

## 7-Consistency Assessment

The consistency of stakeholder judgments is evaluated to verify whether the pairwise comparisons remain logically coherent. Since AHP relies on human judgments, some level of inconsistency is expected. The consistency ratio determines whether the comparison matrix contains acceptable levels of contradiction. The calculation follows three main steps:

**Step 1 — Calculate the Weighted Sum Vector**

The aggregated comparison matrix is multiplied by the priority vector obtained previously. For example, each criterion receives a weighted sum value by combining its comparisons with the calculated criterion weights.

Weighted Sum Vector:

| Criterion | Weighted Sum Value |
|---|---:|
| Financial Efficiency | 0.82 |
| Patient Coverage | 1.82 |
| Operational Robustness | 1.29 |
| Implementation Feasibility | 0.43 |
| Workforce Sustainability | 0.96 |

**Step 2 — Calculate the Consistency Index (CI)**

The weighted sum values are divided by their corresponding priority weights to estimate the maximum eigenvalue (λmax).

Example:

Patient Coverage:

1.82 / 0.34 = 5.35

The same calculation is performed for each criterion:

| Criterion | Weighted Sum / Weight |
|---|---:|
| Financial Efficiency | 5.13 |
| Patient Coverage | 5.35 |
| Operational Robustness | 5.38 |
| Implementation Feasibility | 5.38 |
| Workforce Sustainability | 5.33 |

The average of these values provides the estimated maximum eigenvalue:

λmax ≈ 5.31

The Consistency Index is then calculated as:

CI = (λmax - n) / (n - 1)

Where:

λmax = maximum eigenvalue of the matrix  
n = number of criteria

For this case:

CI = (5.31 - 5) / (5 - 1) = 0.078

**Step 3 — Calculate the Consistency Ratio (CR)**

The Consistency Ratio compares the calculated consistency index with the expected random inconsistency of a matrix of the same size.

CR = CI / RI

Where RI is the Random Index. For a matrix containing 5 criteria:

RI = 1.12

Therefore:

CR = 0.078 / 1.12 = 0.070

**Step 4 — Consistency Ratio Interpretation**

The Consistency Ratio (CR) indicates the degree of coherence between stakeholder judgments. Lower values represent more logically consistent comparisons, while higher values suggest greater inconsistency between pairwise evaluations.

The interpretation scale is:

| Consistency Ratio (CR) | Interpretation |
|---|---|
| CR < 0.05 | Highly consistent judgments. Stakeholders showed strong logical coherence in their comparisons |
| 0.05 ≤ CR < 0.10 | Acceptable consistency. Minor inconsistencies are expected in human judgments and do not affect the reliability of the weights |
| 0.10 ≤ CR < 0.20 | Moderate inconsistency. Results may still be usable, but stakeholder judgments should be reviewed |
| CR ≥ 0.20 | High inconsistency. Pairwise comparisons should be reassessed before using the criteria weights |

For the aggregated stakeholder matrix, the calculated Consistency Ratio (CR) is approximately 0.07. This value remains below the commonly accepted threshold of 0.10, indicating acceptable consistency among stakeholder judgments. Therefore, the resulting criteria weights are considered sufficiently reliable for subsequent TOPSIS-based scenario evaluation.

## 9-Construct the Decision Matrix

The selected workforce planning scenarios are evaluated against the five decision criteria identified during the stakeholder workshop. The resulting decision matrix constitutes the primary input for the subsequent TOPSIS analysis.

- **Financial Efficiency** is measured as the ratio between annual cost savings and total hiring costs, providing a relative indicator of the financial return generated by each workforce planning strategy.

- **Patient Coverage** is measured as the percentage of staffing requirements satisfied under each workforce planning scenario, reflecting the organization's ability to meet expected patient demand.

- **Operational Robustness** reflects the ability of a workforce planning strategy to maintain satisfactory staffing performance when operational conditions differ from the baseline scenario, including variations in patient demand or other planning uncertainties.

- **Implementation Feasibility** reflects the practical ease with which a workforce planning strategy can be implemented, considering recruitment effort, organizational constraints, and operational complexity.

- **Workforce Sustainability** reflects the long-term ability of a workforce planning strategy to support stable staffing levels while maintaining organizational capacity and reducing reliance on temporary workforce solutions.

Operational Robustness, Implementation Feasibility, and Workforce Sustainability are assessed using a five-point expert rating scale ranging from 1 (very low) to 5 (very high). Following the AHP workshop, the same stakeholder panel jointly evaluated each workforce planning scenario and assigned consensus ratings for these qualitative criteria based on the operational characteristics of each proposed strategy.

Although some criteria are naturally related in practice, they capture different decision dimensions rather than duplicate the same information. For example, strategies that improve patient coverage may also become more difficult to implement due to increased recruitment requirements. Likewise, a strategy that is easy to implement may not necessarily provide the highest level of operational robustness or long-term workforce sustainability. Rather than representing redundancy, these relationships reflect the operational trade-offs that hospital decision makers must balance. The objective of the MCDA is therefore to evaluate these competing priorities simultaneously rather than optimize a single performance measure.

The evaluation matrix therefore combines quantitative optimization outputs with qualitative stakeholder assessments, providing a comprehensive representation of the trade-offs associated with each workforce planning scenario.

## 11-Apply TOPSIS Multi-Criteria Ranking

The final decision matrix is used as the input for the TOPSIS multi-criteria decision analysis.

```python
# Load custom library from GitHub
import sys
import pandas as pd
!rm -rf /kaggle/working/jfj-utils
!git clone https://github.com/jfjutras07/jfj-utils.git

# Add to Python path
sys.path.append("/kaggle/working/jfj-utils")
```

```python
from decision.mcda import topsis

decision_matrix = pd.DataFrame(
    {
        "Financial Efficiency": [0.614, 0.611, 0.608, 0.600],
        "Patient Coverage": [70, 75, 80, 85],
        "Operational Robustness": [3, 4, 5, 5],
        "Implementation Feasibility": [5, 4, 3, 2],
        "Workforce Sustainability": [3, 4, 5, 5]
    },
    index=[
        "Budget-Constrained Strategy",
        "Baseline Strategy",
        "Coverage-Oriented Strategy",
        "High-Investment Strategy"
    ]
)

criteria_weights = {
    "Financial Efficiency": 0.16,
    "Patient Coverage": 0.34,
    "Operational Robustness": 0.24,
    "Implementation Feasibility": 0.08,
    "Workforce Sustainability": 0.18
}

benefit_criteria = list(criteria_weights.keys())


topsis_results = topsis(
    df=decision_matrix,
    criteria_weights=criteria_weights,
    benefit_criteria=benefit_criteria
)

topsis_results
```

```python
import matplotlib.pyplot as plt

plot_data = topsis_results.sort_values(
    "TOPSIS_Score",
    ascending=True
)

plt.figure(figsize=(8, 4))

plt.barh(
    plot_data.index,
    plot_data["TOPSIS_Score"]
)

plt.xlabel("TOPSIS Score")
plt.ylabel("Workforce Planning Scenario")
plt.title("TOPSIS Ranking of Workforce Planning Strategies")

plt.show()
```

## 12-TOPSIS Results and Strategic Interpretation

The TOPSIS ranking identifies the Coverage-Oriented Strategy as the preferred workforce planning alternative, with the highest overall score (0.751). This scenario achieves the strongest balance between patient coverage, operational robustness, and workforce sustainability, which are the three criteria receiving the highest stakeholder weights.

The High-Investment Strategy ranks second (0.701). Although it provides the highest patient coverage level, its lower implementation feasibility and reduced financial efficiency limit its overall preference compared with the Coverage-Oriented Strategy.

The Baseline Strategy achieves an intermediate position (0.500), representing a balanced but less competitive alternative across the evaluated criteria.

The Budget-Constrained Strategy obtains the lowest TOPSIS score (0.299). Despite achieving the highest financial efficiency, its lower patient coverage, operational robustness, and workforce sustainability reduce its overall attractiveness when considering multiple organizational priorities simultaneously.

These results illustrate the value of MCDA in strategic workforce planning. The preferred scenario is not the alternative that maximizes a single performance measure, but the one that provides the most balanced compromise across competing hospital objectives.
